## String manipulation

```python
# Example :
# define a string
s = " Hello, World! "

# use upper() method
print(s.upper())  # " HELLO, WORLD! "

# use lower() method
print(s.lower())  # " hello, world! "

# use strip() method
print(s.strip())  # "Hello, World!"

# use split() method
print(s.split(","))  # [' Hello', ' World! ']

# use slicing
print(s[2:5])  # "ell"

# use replace() method
print(s.replace("H", "J"))  # " Jello, World! "

# use rfind() method
print(s.rfind("o"))  # 8
```
