````python
# Small Example:

# Integer list
a=[1,2,3,4,5]
print("Integer list is {}".format(a))

# String list:
b=["banana", "orange", "apple"]
print("String list is {}".format(b))

# String Tuple
t= ("Male", "Female")

# Hybrid list with String , integer, boolean, sub-list and tuple as element
c= ["banana", 1, False, b, t]
print("List with many data types is {}".format(c))
````

````python
# following is the list "a"
a = [12,13, 14, 15, 16]

# you can find the length of list by len() function
print("Lenght of list a {} is {}".format(a, len(a)))
````

````python
# you can extract the particular element from the list like this:
# the counting starts from 0
a= [12,13, 14, 15, 16]

# Second element of list A :
print(a[1])
````

````python
# Suppose Out of 5 elements you only want 3 consecutive elements, we use slicing to get sublist
# Slicing [INBOUND: OUTBOUND]
a= [12,13, 14, 15, 16]

print("From 1st to 3rd element:", a[0:3] )
print("From 2nd to 4th element:", a[1:3])
print("From 2rd to end element:", a[1:]) # we donot define limit
print("From 1st to 4th element:", a[:4]) # we donot define limit
````

````python
# Add another element

a= [12,13, 14, 15, 16]
print("Before appending ", a)
# we use list built in method "append()" to add element to the very last
a.append(17)
print("After appending ", a)
````

````python
# insert another element at certain index

a= [12,13, 14, 15, 16]
print("Before inserting ", a)
# we use list built in method "insert()" to add element to the certain index
a.insert(2,"inserted here")
print("After inserting ", a)
````

````python
# Delete / pop the element at certain position, pop() method returns the element which was popped
a= [12,13, 14, 15, 16]
print("Before popping", a)
a.pop()
print("popped last element", a)
a.pop(2)
print("popped 3rd element",a)
````

````python
# YOUR TURN
from datetime import datetime
# Write a python application, which asks the user for its Student id consisting of 10 characters. Out of those 10 characters last 8 characters denotes the date of birth of the student with XYYYYYMMDD format.
# If the user age is greater than equal to 25 then print "Eligible" otherwise "Ineligible"

STUDENT_ID = input("Enter your student id: ")

if len(STUDENT_ID)==10:
  # YOUR CODE HERE
  dob = STUDENT_ID[-8:]
  dobdate= datetime.strptime(dob, '%Y%m%d')
  todaydate = datetime.today()
  twentyfive_years_ago = datetime(year=(todaydate.year-25), month=todaydate.month, day=todaydate.day)
  if dobdate < twentyfive_years_ago:
    print("Student is Eligible")
  else:
    print("STUDENT is Ineligible")
else:
  print("You have entered incorrect student id . it should contain 10 digits")
````
