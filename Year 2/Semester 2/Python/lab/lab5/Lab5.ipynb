## Python Functions

A Python function is a block of reusable code that performs a specific task when called. Functions are essential for organizing and modularizing code, making it more readable and maintainable. They encapsulate a set of instructions, and you can pass data into them (arguments) and receive data back (return values).

Here's a basic example of how to define and implement a Python function:

```python
def factorial(n):
    """This function calculates the factorial of a non-negative integer."""
    if n == 0:
        return 1
    else:
        return n * factorial(n - 1)

# Calling the function
result = factorial(5)
print(f"The factorial of 5 is {result}")
```

In this example, we have a function named factorial that calculates the factorial of a non-negative integer n. The factorial of a non-negative integer n is the product of all positive integers from 1 to n.

Here's how this function works:

The function takes one argument, n, representing the number for which you want to calculate the factorial.

It has a base case: if n == 0, the function returns 1. This is necessary to stop the recursion and prevent infinite recursion when n reaches 0.

If n is not 0, the function recursively calls itself with n - 1 and multiplies the result by n. This process continues until the base case is reached, and then it "unwinds" with each result being multiplied by the next n until it reaches the original n.

Finally, the function is called with factorial(5), and the result is stored in the variable result. The result is then printed to the console.

This is an example of a more complex function that uses recursion to solve a mathematical problem. It showcases the power of functions in Python for breaking down complex tasks into smaller, more manageable parts.

```python
# Here is an Example of Implemented Stack Datastructure which uses LIFO (Last In First Out)
class Stack:
  def __init__(self):
    self.__stk = []
  def push(self, val):
    self.__stk.append(val)
  def pop(self):
    val = self.__stk[-1]
    del self.__stk[-1]
    return val
  def print_stack(self):
    print(self.__stk)

stack = Stack()
stack.print_stack()
stack.push(5)
stack.push(1)
stack.print_stack()
stack.push(6)
stack.print_stack()
popped_element = stack.pop()
stack.print_stack()
print("Popped Element :", popped_element)

```

```python
# YOUR TURN:
# Implement the Queue Data structure. It should use FIFO (First In First Out).
# Methods should be push, pop, print_queue

class Queue:
  def __init__(self):
    self.__stk = []
  def enqueue(self, val):
    self.__stk.append(val)
  def dequeue(self):
    val = self.__stk[0]
    del self.__stk[0]
    return val
  def print_queue(self):
    print(self.__stk)

q = Queue()
q.enqueue(3)
q.enqueue(5)
q.enqueue(10)
q.print_queue()
q.dequeue()
q.print_queue()
popped_element = q.pop()
print(f"Popped element {popped_element}")
```
